# Task 3: Recommendation System Analysis — Netflix Dataset

## Objective

The objective of this task is to build a simple content-based recommendation system using the Netflix dataset.

The recommendation system will:
- Use Netflix content categories to represent each title.
- Preprocess the text data.
- Convert categories into numerical features using TF-IDF.
- Calculate similarity between titles using cosine similarity.
- Recommend titles that are similar to a selected title.

This approach is called a **content-based recommendation system** because recommendations are based on the characteristics of the content.

## 1. Preparing Content Features

For a content-based recommendation system, we need information that describes each title.

The Netflix dataset does not contain a description field, so we will use the `listed_in` column, which contains the categories and genres associated with each title.

These categories will be used as the content features for calculating similarity between titles.

```python
# Combine content type, rating, and categories
df["content_features"] = (
    df["type"].fillna("") + " " +
    df["rating"].fillna("") + " " +
    df["listed_in"].fillna("")
).str.lower()

# Display sample titles and their content features
df[["title", "content_features"]].head(10)
```

## 2. Text Preprocessing and TF-IDF

The category information is text-based, so it needs to be converted into numerical values before similarity can be calculated.

We will use **TF-IDF (Term Frequency-Inverse Document Frequency)** to represent the categories numerically.

TF-IDF gives higher importance to terms that are useful for distinguishing between different titles.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Create the TF-IDF vectorizer
tfidf = TfidfVectorizer()

# Convert content features into TF-IDF vectors
tfidf_matrix = tfidf.fit_transform(df["content_features"])

# Check the shape of the TF-IDF matrix
tfidf_matrix.shape
```

## 3. Calculating Similarity

After converting the Netflix categories into TF-IDF vectors, we can calculate how similar each title is to every other title.

We will use **cosine similarity** to measure the similarity between the TF-IDF vectors.

A higher cosine similarity score means that two titles have more similar categories.

```python
from sklearn.metrics.pairwise import cosine_similarity

# Calculate similarity between all Netflix titles
similarity_matrix = cosine_similarity(tfidf_matrix)

# Check the shape of the similarity matrix
similarity_matrix.shape
```

## 4. Building the Recommendation Function

We will create a recommendation function that takes a Netflix title as input and returns the most similar titles.

The function will:
1. Find the selected title in the dataset.
2. Retrieve its similarity scores.
3. Sort the titles by similarity.
4. Return the top 10 most similar titles.

```python
def recommend_titles(title, num_recommendations=10):
    # Find the index of the selected title
    title_index = df[df["title"].str.lower() == title.lower()].index[0]

    # Get similarity scores for the selected title
    similarity_scores = list(enumerate(similarity_matrix[title_index]))

    # Sort titles by similarity score
    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    # Get the top recommendations, excluding the selected title itself
    recommendations = similarity_scores[1:num_recommendations + 1]

    # Get the title names and similarity scores
    results = []

    for index, score in recommendations:
        results.append({
            "Title": df.iloc[index]["title"],
            "Similarity Score": round(score, 3)
        })

    return pd.DataFrame(results)
```

```python
recommend_titles("The Starling")
```

## 5. Recommendation Quality Evaluation

The recommendation system was evaluated by testing it with different Netflix titles and examining whether the recommended titles shared similar content characteristics.

A recommendation is considered more relevant when the recommended title has similar:
- Content type
- Rating
- Categories or genres

The similarity score provides a numerical measure of how closely the recommended title matches the selected title.

Higher similarity scores indicate stronger content-based similarity.

```python
# Test the recommendation system with another Netflix title
recommend_titles("Midnight Mass")
```

```python
recommend_titles("The Perks of Being a Wallflower")
```

## 6. Evaluation Findings

The recommendation system was tested using multiple Netflix titles, including *The Starling*, *Midnight Mass*, and *The Perks of Being a Wallflower*.

The results showed that the system generally recommended titles with similar content types, ratings, and categories.

For example:
- *The Starling* produced recommendations with high similarity scores, including titles categorized as Comedies and Dramas.
- *Midnight Mass* produced recommendations related to horror, drama, and similar TV content.
- *The Perks of Being a Wallflower* produced recommendations with similar Movie, rating, and category features.

Several recommendations received a similarity score of **1.000** because they had the same feature combination used by the model. This demonstrates that the system successfully identifies titles with matching content characteristics, but it also shows a limitation of using a small set of metadata features.

Overall, the recommendation system provides reasonable content-based recommendations using TF-IDF and cosine similarity.

# Task 5: Machine Learning Classification Model — Netflix Dataset

## Objective

The objective of this task is to build a machine learning classification model that predicts the type of Netflix content based on available features.

In this task, we will:
- Select relevant features from the dataset.
- Prepare the data for machine learning.
- Split the data into training and testing sets.
- Train classification models.
- Evaluate model performance.
- Compare the accuracy of different models.

```python
# Select features for classification

features = df[[
    "release_year",
    "rating"
]].copy()

target = df["type"]

# Display the selected features
features.head()
```

```python
# Convert the categorical rating feature into numerical values

features = pd.get_dummies(
    features,
    columns=["rating"],
    drop_first=True
)

# Display the prepared features
features.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    features,
    target,
    test_size=0.2,
    random_state=42,
    stratify=target
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression

# Create and train the model
model = LogisticRegression(max_iter=5000)

model.fit(X_train, y_train)

print("Model training completed.")
```

```python
from sklearn.metrics import accuracy_score, classification_report

# Make predictions on the test data
y_pred = model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", round(accuracy, 4))
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
from sklearn.tree import DecisionTreeClassifier

# Create and train the Decision Tree model
tree_model = DecisionTreeClassifier(
    random_state=42,
    max_depth=5
)

tree_model.fit(X_train, y_train)

print("Decision Tree training completed.")
```

```python
# Make predictions using the Decision Tree
tree_pred = tree_model.predict(X_test)

# Calculate accuracy
tree_accuracy = accuracy_score(y_test, tree_pred)

print("Decision Tree Accuracy:", round(tree_accuracy, 4))

print("\nClassification Report:")
print(classification_report(y_test, tree_pred))
```

```python
# Compare model performance

model_comparison = pd.DataFrame({
    "Model": ["Logistic Regression", "Decision Tree"],
    "Accuracy": [accuracy, tree_accuracy]
})

model_comparison
```

## 6. Model Comparison and Findings

Two classification models were trained and evaluated: Logistic Regression and Decision Tree.

| Model | Accuracy |
|---|---:|
| Logistic Regression | 70.65% |
| Decision Tree | 71.79% |

The Decision Tree achieved the higher accuracy of **71.79%**, compared with **70.65%** for Logistic Regression.

The classification reports also showed that both models performed better at identifying Movies than TV Shows. The Decision Tree improved the recall for TV Shows compared with Logistic Regression.

Overall, the Decision Tree performed slightly better on this dataset and was selected as the better-performing model based on accuracy.

```python
# Visual comparison of model accuracy

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.bar(
    model_comparison["Model"],
    model_comparison["Accuracy"]
)

plt.title("Classification Model Accuracy Comparison")
plt.xlabel("Model")
plt.ylabel("Accuracy")
plt.ylim(0, 1)

plt.show()
```
